## Splitting data into train test split before-hand to avoild Data Leakage

- Here data was splitted before augmentation to avoid data leakage

```python
import cv2 as cv
import os
import random

src_path = '.\\Raw DataSet\\Diseased'

train_dst = '.\\Augumented DataSet\\Train\\Diseased'
test_dst  = '.\\Augumented DataSet\\Test\\Diseased'

os.makedirs(train_dst, exist_ok=True)
os.makedirs(test_dst, exist_ok=True)

def save_image(img, prefix, filename, dst_path):
    """Save an augmented image to the destination folder."""
    cv.imwrite(os.path.join(dst_path, prefix + filename), img)

def augment_and_save(file_name, dst_path):
    full_path = os.path.join(src_path, file_name)
    img = cv.imread(full_path)
    if img is None:
        return

    save_image(img, 'a0_', file_name, dst_path)

    # Flips: horizontal, vertical, both
    save_image(cv.flip(img, 1),  'a1_', file_name, dst_path)
    save_image(cv.flip(img, 0),  'a2_', file_name, dst_path)
    save_image(cv.flip(img, -1), 'a3_', file_name, dst_path)

    # Rotations: 90 CW, 180, 90 CCW
    save_image(cv.rotate(img, cv.ROTATE_90_CLOCKWISE),        'r1_', file_name, dst_path)
    save_image(cv.rotate(img, cv.ROTATE_180),                 'r2_', file_name, dst_path)
    save_image(cv.rotate(img, cv.ROTATE_90_COUNTERCLOCKWISE), 'r3_', file_name, dst_path)

# spliting ORIGINAL filenames before any augmentation 
all_files = [f for f in os.listdir(src_path) if os.path.isfile(os.path.join(src_path, f))]

random.seed(42)          
random.shuffle(all_files)

split_idx = int(0.8 * len(all_files))   # 80% - 20% split
train_files = all_files[:split_idx]
test_files  = all_files[split_idx:]

print(f'Total originals: {len(all_files)} -> Train: {len(train_files)}, Test: {len(test_files)}')


for file_name in train_files:
    augment_and_save(file_name, train_dst)

for file_name in test_files:
    augment_and_save(file_name, test_dst)

print('Augmentation with leak-free split successful')
```
